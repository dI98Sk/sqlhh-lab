# hh_lab — подключение к базе

Сервер: Homebrew PostgreSQL 16, `localhost:5433`, база `hh_lab`, вход без пароля.

```python
import getpass
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine(f"postgresql+psycopg2://{getpass.getuser()}@localhost:5433/hh_lab")

def q(sql, **params):
    """Выполнить SELECT и вернуть DataFrame."""
    with engine.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

## Размер таблиц

```python
q("""
SELECT relname AS table_name,
       n_live_tup AS rows,
       pg_size_pretty(pg_total_relation_size(relid)) AS size
FROM pg_stat_user_tables
ORDER BY pg_total_relation_size(relid) DESC
""")
```

## Пример:


```python
q("""
SELECT s.name, count(*) AS vacancies
FROM vacancy_skills vs
JOIN skills s USING (skill_id)
GROUP BY s.name
ORDER BY vacancies DESC
LIMIT 15
""")
```
